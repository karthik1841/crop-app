```python
import cv2
import numpy as np
from ultralytics import YOLO

# Load the YOLOv8 model
model = YOLO("runs/detect/tomato_disease_model2/weights/best.pt")  # Use best.pt (mAP50=83.3%)

# Load the image
image_path = "/content/images.jpg"
img = cv2.imread(image_path)
if img is None:
    raise FileNotFoundError(f"Image not found at {image_path}")

# Perform inference
results = model.predict(image_path, save=False, conf=0.1)  # conf=0.1 to detect weak classes

# Process results
for result in results:
    boxes = result.boxes  # Bounding box data
    for box in boxes:
        # Extract coordinates, confidence, and class ID
        x1, y1, x2, y2 = box.xyxy[0].cpu().numpy().astype(int)  # Convert tensor to numpy
        confidence = box.conf.item()  # Convert tensor to float
        class_id = int(box.cls.item())  # Convert tensor to int
        class_name = model.names[class_id]  # Get class name (e.g., 'Septoria')

        # Draw bounding box
        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

        # Add label with confidence
        label = f"{class_name}: {confidence:.2f}"
        cv2.putText(img, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

# Save and display the result
output_path = "/content/download_annotated1.jpg"
cv2.imwrite(output_path, img)
print(f"Annotated image saved to {output_path}")

# Display image (optional, for Colab)
from google.colab.patches import cv2_imshow
cv2_imshow(img)
```
